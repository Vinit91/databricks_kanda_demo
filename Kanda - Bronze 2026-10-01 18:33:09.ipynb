```python
import re

# 1. billable_hours path
volume_file_path = '/Volumes/workspace/default/vinitvolume/billable_hours.csv'

# 2. Read the CSV file exactly as-is into Spark
billable_hours_raw = (spark.read
    .format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load(volume_file_path)
)

# 3. Clean column names (replace spaces and special characters with underscores)
for col_name in billable_hours_raw.columns:
    # Convert spaces/hyphens to underscores and remove punctuation
    clean_name = re.sub(r'[ ,;{}()\n\t=]', '_', col_name)
    billable_hours_raw = billable_hours_raw.withColumnRenamed(col_name, clean_name)

# 4. Save safely to your Bronze Schema as a Delta Table
billable_hours_raw.write.format("delta").mode("append").saveAsTable("workspace.bronze.billable_hours_raw")

print("billable_hours_raw table successfully populated from our uploaded sample file containing billable hours data!")
```
